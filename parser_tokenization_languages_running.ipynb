```python
import pickle
with open("/content/stanza_t_languages.pickle", 'rb') as f:
    data = pickle.load(f)
```

```python
data.keys()
```

```python
import json

for lang in data.keys():
  filename = f'stanza_t_{lang}.jsonl'
  with open(filename, 'w', encoding='utf-8') as json_file:
    for d_i, d in enumerate(data[lang]):
      d_dict = {'index': d_i, 'input': d.text,
        'pred_tree': [{'id': t.id, 'form': t.form,
          'parent_id': t.parent_id, 'relation': t.relation} for t in d.tokens]}
      json_file.write(json.dumps(d_dict, ensure_ascii=False) + '\n')
    json_file.flush()
  files.download(filename)
```
